## 6. Data Cleaning

### 6.1 Create a Working Copy

Create a copy of the original dataset to perform cleaning operations while preserving the raw data.

```python
df_clean = df.copy()
df_clean.shape
```

### 6.2 Select Relevant Columns

Select the columns needed to analyze developer salaries, work experience, job roles, countries, education, and programming languages.

```python
selected_columns = [
    'Country',
    'DevType',
    'Employment',
    'WorkExp',
    'YearsCode',
    'EdLevel',
    'ConvertedCompYearly',
    'LanguageHaveWorkedWith',
    'RemoteWork'
]

df_clean = df_clean[selected_columns].copy()

df_clean.shape
```

### 6.3 Missing Values Analysis

Examine the number and percentage of missing values in the selected columns before deciding how to handle them.

```python
missing_values = pd.DataFrame({
    'Missing Count': df_clean.isna().sum(),
    'Missing Percentage': df_clean.isna().mean() * 100
})

missing_values.round(2)
```

### 6.4 Prepare Salary Data

Create a separate dataset containing respondents with reported annual compensation. Missing salary values are excluded from salary analysis without modifying the original dataset.

```python
salary_df = df_clean.dropna(subset=['ConvertedCompYearly']).copy()

salary_df.shape
```

### 6.5 Salary Distribution and Outlier Detection

Examine the distribution of annual compensation to identify unusually low or high values before applying any filtering.

```python
salary_df['ConvertedCompYearly'].describe(
    percentiles=[0.01, 0.25, 0.50, 0.75, 0.95, 0.99]
)
```

### 6.6 Examine Extreme Salary Values

Count unusually low and high annual compensation values to assess potential outliers before deciding whether filtering is appropriate.

```python
print("Below $1,000:", (salary_df['ConvertedCompYearly'] < 1000).sum())

print("Above $1,000,000:", (salary_df['ConvertedCompYearly'] > 1000000).sum())
```

### 6.7 Create a Filtered Salary Dataset

Create a separate salary dataset using an exploratory annual compensation range of $1,000 to $1,000,000. Preserve the original salary data for comparison and sensitivity analysis.

```python
salary_filtered = salary_df[
    salary_df['ConvertedCompYearly'].between(1000, 1000000)
].copy()

print("Before filtering:", len(salary_df))
print("After filtering:", len(salary_filtered))
print("Excluded records:", len(salary_df) - len(salary_filtered))
```

### 6.8 Compare Salary Statistics Before and After Filtering

Compare the mean and median annual compensation before and after filtering to evaluate the impact of extreme values.

```python
comparison = pd.DataFrame({
    'Before Filtering': [
        salary_df['ConvertedCompYearly'].mean(),
        salary_df['ConvertedCompYearly'].median()
    ],
    'After Filtering': [
        salary_filtered['ConvertedCompYearly'].mean(),
        salary_filtered['ConvertedCompYearly'].median()
    ]
}, index=['Mean', 'Median'])

comparison.round(2)
```

## 7. Exploratory Data Analysis (EDA)

### 7.1 Salary vs Work Experience

Analyze the relationship between years of work experience and median annual compensation.

```python
experience_df = salary_filtered.dropna(subset=['WorkExp']).copy()

experience_salary = (
    experience_df.groupby('WorkExp')['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
    .sort_values('WorkExp')
)

experience_salary.head(10)
```

### 7.2 Visualizing Salary by Work Experience

Use a line chart to visualize how median annual compensation varies across years of work experience.

```python
import plotly.express as px

fig = px.line(
    experience_salary,
    x='WorkExp',
    y='median',
    markers=True,
    title='Median Annual Salary by Work Experience',
    labels={
        'WorkExp': 'Years of Work Experience',
        'median': 'Median Annual Salary (USD)'
    }
)

fig.show()
```

**Insight:** Median annual compensation generally increases with work experience in the first ten years. However, salary differences may also depend on country and job role.

### 7.3 Salary by Job Role

Compare median annual compensation across different developer roles to identify salary differences between career paths.

```python
role_df = salary_filtered.dropna(subset=['DevType']).copy()

role_salary = (
    role_df.groupby('DevType')['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
    .sort_values('median', ascending=False)
)

role_salary.head(15)
```

### 7.4 Visualizing Salary by Job Role

Visualize the top 15 job roles by median annual compensation using a horizontal bar chart.

```python
top_roles = role_salary.head(15)

fig = px.bar(
    top_roles,
    x='median',
    y='DevType',
    orientation='h',
    title='Top 15 Job Roles by Median Annual Salary',
    labels={
        'median': 'Median Annual Salary (USD)',
        'DevType': 'Job Role'
    },
    hover_data=['count']
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    height=650
)

fig.show()
```

**Insight:** Senior executives and engineering managers have the highest median annual compensation among the displayed roles. Salary differences across roles may also reflect variations in experience, location, and other factors.

### 7.5 Salary by Country

Compare median annual compensation across countries. Include the number of respondents in each country to avoid relying on very small samples.

```python
country_df = salary_filtered.dropna(subset=['Country']).copy()

country_salary = (
    country_df.groupby('Country')['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
    .sort_values('median', ascending=False)
)

country_salary.head(15)
```

### 7.6 Filter Countries by Sample Size

Include only countries with at least 100 respondents reporting annual compensation to reduce the influence of very small samples.

```python
country_salary_filtered = country_salary[
    country_salary['count'] >= 100
].copy()

country_salary_filtered.head(15)
```

### 7.7 Visualizing Salary by Country

Visualize the top 15 countries by median annual compensation, considering only countries with at least 100 salary responses.

```python
top_countries = country_salary_filtered.head(15)

fig = px.bar(
    top_countries,
    x='median',
    y='Country',
    orientation='h',
    title='Top 15 Countries by Median Annual Salary',
    labels={
        'median': 'Median Annual Salary (USD)',
        'Country': 'Country'
    },
    hover_data=['count']
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    height=650,
    margin=dict(l=230)
)

fig.show()
```

**Insight:** The United States has the highest median annual compensation ($150,000) among countries with at least 100 salary responses, followed by Switzerland ($142,592) and Israel ($141,188).

These differences may reflect variations in local economies, cost of living, job roles, and experience levels. The results describe survey respondents and should not be interpreted as representative national salary averages.

### 7.8 Most Popular Programming Languages

Analyze the programming languages respondents have worked with to identify the most commonly reported technologies in the survey.

```python
languages = df_clean['LanguageHaveWorkedWith'].dropna()

language_counts = (
    languages
    .str.split(';')
    .explode()
    .str.strip()
    .value_counts()
)

language_counts.head(15)
```

### 7.9 Visualizing the Most Popular Programming Languages

Visualize the 15 most commonly reported programming languages and technologies among survey respondents.

```python
top_languages = language_counts.head(15).reset_index()
top_languages.columns = ['Language', 'Count']

fig = px.bar(
    top_languages,
    x='Count',
    y='Language',
    orientation='h',
    title='Top 15 Most Popular Programming Languages',
    labels={
        'Count': 'Number of Respondents',
        'Language': 'Programming Language / Technology'
    },
    text='Count'
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    height=650
)

fig.show()
```

**Insight:** JavaScript is the most commonly reported technology (21,005 respondents), followed by HTML/CSS (19,698), SQL (18,633), and Python (18,410).

These results highlight the widespread use of web development technologies, databases, and Python among survey respondents. Since respondents could select multiple technologies, the counts are not mutually exclusive.

### 7.10 Salary by Education Level

Analyze median annual compensation across different education levels to explore the relationship between educational attainment and developer salaries.

```python
education_df = salary_filtered.dropna(subset=['EdLevel']).copy()

education_salary = (
    education_df.groupby('EdLevel')['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
    .sort_values('median', ascending=False)
)

education_salary
```

### 7.11 Visualizing Salary by Education Level

Compare median annual compensation across education levels using a horizontal bar chart.

```python
education_plot = education_salary.copy()

education_plot['Education'] = education_plot['EdLevel'].replace({
    'Professional degree (JD, MD, Ph.D, Ed.D, etc.)': 'Professional Degree',
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)': "Master's Degree",
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)': "Bachelor's Degree",
    'Associate degree (A.A., A.S., etc.)': 'Associate Degree',
    'Primary/elementary school': 'Primary School',
    'Other (please specify):': 'Other'
})

fig = px.bar(
    education_plot,
    x='median',
    y='Education',
    orientation='h',
    title='Median Annual Salary by Education Level',
    labels={
        'median': 'Median Annual Salary (USD)',
        'Education': 'Education Level'
    },
    hover_data=['count']
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    height=550,
    margin=dict(l=280)
)

fig.show()
```

**Insight:** Respondents with professional degrees report the highest median annual compensation ($89,257), followed by master's degree holders ($81,210) and bachelor's degree holders ($80,000).

Although higher educational attainment is associated with higher median compensation in several groups, these differences do not establish a causal relationship. Work experience, country, and job role may also influence salary.

### 7.12 Salary by Work Arrangement

Analyze median annual compensation across different work arrangements, including remote, hybrid, and in-person work.

```python
remote_df = salary_filtered.dropna(subset=['RemoteWork']).copy()

remote_salary = (
    remote_df.groupby('RemoteWork')['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
    .sort_values('median', ascending=False)
)

remote_salary
```

### 7.13 Visualizing Salary by Work Arrangement

Compare median annual compensation across different work arrangements using a bar chart.

```python
remote_plot = remote_salary.copy()

remote_plot['Work Arrangement'] = [
    'Remote',
    'Hybrid - Flexible',
    'Your Choice',
    'Hybrid - In-person',
    'In-person'
]

fig = px.bar(
    remote_plot,
    x='Work Arrangement',
    y='median',
    title='Median Annual Salary by Work Arrangement',
    labels={
        'median': 'Median Annual Salary (USD)',
        'Work Arrangement': 'Work Arrangement'
    },
    text='median',
    hover_data=['count']
)

fig.update_traces(texttemplate='$%{text:,.0f}', textposition='outside')

fig.update_layout(
    height=500,
    yaxis_range=[0, 110000]
)

fig.show()
```

**Insight:** Remote workers report the highest median annual compensation ($91,198), while in-person workers report the lowest ($48,346.5).

The observed difference is approximately $42,852 annually. However, this is an association rather than evidence that remote work causes higher salaries. Differences in country, job role, and experience may contribute to this gap.

## 8. Statistical Analysis

### 8.1 Correlation Between Work Experience and Salary

Use Spearman's rank correlation coefficient to measure the strength and direction of the relationship between years of work experience and annual compensation.

```python
correlation_df = salary_filtered.dropna(
    subset=['WorkExp', 'ConvertedCompYearly']
).copy()

correlation = correlation_df[
    'WorkExp'
].corr(
    correlation_df['ConvertedCompYearly'],
    method='spearman'
)

print("Spearman Correlation:", round(correlation, 3))
print("Number of respondents:", len(correlation_df))
```

**Statistical Insight:** Spearman's rank correlation coefficient between work experience and annual compensation is 0.44, based on 22,769 respondents.

This indicates a moderate positive association: respondents with more work experience tend to report higher annual salaries. However, correlation does not imply causation, and other factors such as country and job role may influence compensation.

### 8.2 Salary Comparison Across Experience Groups

Group respondents by years of professional experience and compare median annual compensation across career stages.

```python
experience_groups = correlation_df.copy()

experience_groups['Experience Group'] = pd.cut(
    experience_groups['WorkExp'],
    bins=[-1, 2, 5, 10, 20, float('inf')],
    labels=[
        '0-2 years',
        '3-5 years',
        '6-10 years',
        '11-20 years',
        '20+ years'
    ]
)

experience_group_salary = (
    experience_groups.groupby(
        'Experience Group', observed=True
    )['ConvertedCompYearly']
    .agg(['median', 'count'])
    .reset_index()
)

experience_group_salary
```

### 8.3 Visualizing Salary Across Experience Groups

Visualize median annual compensation across experience groups to compare salary patterns at different career stages.

```python
fig = px.bar(
    experience_group_salary,
    x='Experience Group',
    y='median',
    title='Median Annual Salary by Experience Group',
    labels={
        'Experience Group': 'Years of Work Experience',
        'median': 'Median Annual Salary (USD)'
    },
    text='median',
    hover_data=['count']
)

fig.update_traces(
    texttemplate='$%{text:,.0f}',
    textposition='outside'
)

fig.update_layout(
    height=500,
    yaxis_range=[0, 130000]
)

fig.show()
```

**Insight:** Median annual compensation increases across all five experience groups, from $25,523 for respondents with 0–2 years of experience to $110,214 for those with more than 20 years.

This pattern supports the moderate positive Spearman correlation (0.44) observed earlier. However, the results are descriptive and may also reflect differences in country, job role, and other factors.
